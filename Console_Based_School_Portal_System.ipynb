## Implement multi-student input loop

### Subtask:
Create a loop that repeatedly calls the `add_student` function, allowing the user to add multiple students until they choose to stop.


**Reasoning**:
To allow the user to add multiple students, I need to create a `while` loop that repeatedly calls the `add_student` function and provides an option to exit the loop.



```python
students = [] # Initialize the students list

def add_student():
    print("Add Student ")
    sid = input("Enter Student ID: ")
    name = input("Enter Student Name: ")
    age = input("Enter Student Age: ")
    student_class = input("Enter Class: ")

    subjects_marks = []
    while True:
        subject_name = input("Enter Subject Name (or 'done' to finish): ")
        if subject_name.lower() == 'done':
            break
        marks = input(f"Enter marks for {subject_name}: ")
        subjects_marks.append({"subject": subject_name, "marks": marks})

    student = {
        "id": sid,
        "name": name,
        "age": age,
        "class": student_class,
        "subjects": subjects_marks
    }

    students.append(student)
    print("Student added successfully!\n")

# Loop to add multiple students
while True:
    add_student()
    add_more = input("Do you want to add another student? (yes/no): ")
    if add_more.lower() == 'no':
        break

print("Student management system updated. Final student list:")
for student in students:
    print(student)
```

### Finding a student by ID

This code block demonstrates how to find a student in the `students` list by their ID. It iterates through each student dictionary and checks if the 'id' field matches the `search_id`.

```python
search_id = input("Enter the ID of the student you want to find: ")
found_student = None

for student in students:
    if student['id'] == search_id:
        found_student = student
        break # Exit the loop once the student is found

if found_student:
    print(f"\nStudent Found for ID '{search_id}':")
    print(found_student)
else:
    print(f"\nNo student found with ID '{search_id}'.")
```

## Calculate Average Grades per Student

### Subtask:
Develop a function to calculate the average marks for each student across all their subjects. This will involve iterating through the 'subjects' list for each student and averaging their 'marks'.


**Reasoning**:
To address the subtask, I need to create a Python function that iterates through the list of students, calculates the average marks for each student based on their subjects, and then updates each student's dictionary with this average. I will also call this function immediately after its definition to apply it to the existing 'students' list.



```python
def calculate_average_grades(students_list):
    for student in students_list:
        total_marks = 0
        subject_count = 0

        # Ensure 'subjects' key exists and is a list
        if 'subjects' in student and isinstance(student['subjects'], list):
            for subject_info in student['subjects']:
                try:
                    # Convert marks to float before adding
                    marks = float(subject_info['marks'])
                    total_marks += marks
                    subject_count += 1
                except (ValueError, TypeError):
                    print(f"Warning: Invalid marks found for subject '{subject_info.get('subject', 'N/A')}' for student '{student.get('name', 'N/A')}'. Skipping these marks.")

        # Calculate average marks
        if subject_count > 0:
            average_marks = total_marks / subject_count
        else:
            average_marks = 0 # Assign 0 if no valid subjects/marks

        # Add 'average_grade' key to the student dictionary
        student['average_grade'] = round(average_marks, 2) # Round to two decimal places

    print("Average grades calculated and added for all students.\n")

# Call the function with the global students list
calculate_average_grades(students)

# Display updated student list to confirm
print("--- Students with Average Grades ---")
if not students:
    print("No students have been added yet.")
else:
    for student in students:
        print(student)
print("----------------------------------")
```

```python
print("\n--- Analyzing Grade Distribution ---")

# Define grade boundaries
grade_boundaries = {
    'A': 90,
    'B': 80,
    'C': 70,
    'D': 60,
    'F': 0
}

# Initialize grade counts
grade_counts = {'A': 0, 'B': 0, 'C': 0, 'D': 0, 'F': 0}

if not students:
    print("No students available to analyze grade distribution.")
else:
    for student in students:
        average_grade = student.get('average_grade', 0) # Get average_grade, default to 0 if not found

        if average_grade >= grade_boundaries['A']:
            grade_counts['A'] += 1
        elif average_grade >= grade_boundaries['B']:
            grade_counts['B'] += 1
        elif average_grade >= grade_boundaries['C']:
            grade_counts['C'] += 1
        elif average_grade >= grade_boundaries['D']:
            grade_counts['D'] += 1
        else:
            grade_counts['F'] += 1

    print("\nGrade Distribution:")
    for grade, count in grade_counts.items():
        print(f"Grade {grade}: {count} student(s)")

print("------------------------------------")
```

**Reasoning**:

```python
print("\n--- Identifying Top Performing Students ---")

if not students:
    print("No students available to identify top performers.")
else:
    # Sort students by 'average_grade' in descending order
    sorted_students = sorted(students, key=lambda s: s.get('average_grade', 0), reverse=True)

    # Determine how many top performers to display
    num_top_performers = min(3, len(sorted_students)) # Display top 3 or fewer if less than 3 students

    if num_top_performers > 0:
        print(f"\nTop {num_top_performers} Performing Student(s):")
        for i in range(num_top_performers):
            student = sorted_students[i]
            print(f"  Name: {student['name']}, Average Grade: {student.get('average_grade', 'N/A')}")
    else:
        print("No students with calculated average grades to identify top performers.")

print("------------------------------------------")
```
